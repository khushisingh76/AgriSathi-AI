```python
from ultralytics import YOLO

model = YOLO(
    r"runs/detect/AgriSathi_YOLO-3/weights/best.pt"
)

print("AgriSathi model loaded successfully!")
```

```python
from IPython.display import Image, display

image_path = r"C:\Users\Khushi singh\Downloads\test_leaf.jpg"

results = model.predict(
    source=image_path,
    imgsz=416,
    conf=0.25,
    save=True
)

display(Image(filename=results[0].save_dir + 
              "/" + results[0].path.split("\\")[-1]))
```

```python
results = model.predict(
    source=image_path,
    imgsz=416,
    conf=0.50,
    iou=0.45,
    save=True
)

for result in results:
    for box in result.boxes:
        class_id = int(box.cls[0])
        confidence = float(box.conf[0])

        print(
            "Prediction:", result.names[class_id],
            "| Confidence:", round(confidence * 100, 2), "%"
        )
```

```python
results = model.predict(
    source=image_path,
    imgsz=416,
    conf=0.50,
    iou=0.30,
    max_det=10,
    save=True
)

for result in results:
    print("Total detections:", len(result.boxes))

    for box in result.boxes:
        class_id = int(box.cls[0])
        confidence = float(box.conf[0])

        print(
            result.names[class_id],
            round(confidence * 100, 2), "%"
        )
```
